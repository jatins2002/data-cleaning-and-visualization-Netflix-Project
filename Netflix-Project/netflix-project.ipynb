```python
#clean the data

#-----how many movies and tv shows are there on netflix------
df = df.dropna(subset=['type','release_year','rating','country','duration'])

type_counts = df['type'].value_counts()
plt.figure(figsize=(6,4))
plt.bar(type_counts.index,type_counts.values,color=['skyblue','orange'])
plt.title('Number of Movies and TV Shows on Netflix')
plt.xlabel('Content Type')
plt.ylabel('Count')
plt.tight_layout()
plt.savefig('content_type_distribution.png')
plt.legend()
plt.show()

```

```python
#-----percentage of the ratings on netflix------

ratings_counts = df['rating'].value_counts()
colour = ['skyblue','orange','green','red','purple']
plt.figure(figsize=(10,6))
plt.pie(ratings_counts.values, labels=ratings_counts.index, autopct='%1.1f%%', colors=colour)
plt.title('Percentage of Distribution of Content Ratings on Netflix')
plt.tight_layout()
plt.savefig('rating_distribution.png')
plt.show()
```

```python
#--------distribution of movie durations------

movie_df = df[df['type'] == 'Movie'].copy()
movie_df['duration_int'] = movie_df['duration'].str.replace(' min','').astype(int)

plt.figure(figsize=(8,6))
plt.hist(movie_df['duration_int'],bins=20,edgecolor='black')
plt.title('Distribution of Movie Durations on Netflix')
plt.xlabel('Duration (minutes)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.savefig('movie_duration_distribution.png')
plt.show()


```

```python
#------relations b/w release year and no of shows------

release_counts = df['release_year'].value_counts().sort_index()
plt.figure(figsize=(10,6))
plt.scatter(release_counts.index,release_counts.values,color='red')
plt.title('Number of Shows on Netflix by Release Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Shows')
plt.tight_layout()
plt.savefig('release_year_distribution.png')
plt.show()
```

```python
#-----top 10 countries with the most content on netflix------

country_counts = df['country'].value_counts().head(10)
plt.figure(figsize=(10,6))
plt.barh(country_counts.index,country_counts.values,color='skyblue')
plt.title('Top 10 Countries with the Most Content on Netflix')
plt.xlabel('Country')
plt.ylabel('Number of Shows')
plt.tight_layout()
plt.savefig('country_content_distribution.png')
plt.show()
```

```python
#-----compare content by year------

content_by_year = df.groupby(['release_year','type']).size().unstack().fillna(0)

fig, ax = plt.subplots(1,2,figsize=(12,5))
fig.suptitle('Content Distribution by Year')

#first --> subplot for movies
ax[0].plot(content_by_year.index,content_by_year['Movie'],label='Movies',marker='o',linestyle='--',color='blue')
ax[0].set_title('Number of Movies on Netflix by Year')
ax[0].set_xlabel('Release Year')
ax[0].set_ylabel('Number of Movies')
ax[0].legend()
ax[0].grid(True)

#second --> subplot for tv shows
ax[1].plot(content_by_year.index,content_by_year['TV Show'],label='TV Shows',marker='o',linestyle='--',color='orange')
ax[1].set_title('Number of TV Shows on Netflix by Year')
ax[1].set_xlabel('Release Year')
ax[1].set_ylabel('Number of TV Shows')
ax[1].legend()
ax[1].grid(True)
plt.tight_layout()
plt.savefig('content_by_year.png')
plt.show()
```

```python
#-----compare content by year------

content_by_year = df.groupby(['release_year','type']).size().unstack().fillna(0)

fig, ax = plt.subplots(1,1,figsize=(12,5))
fig.suptitle('Content Distribution(Movies & TV Shows) by Year')

#first --> subplot for movies
ax.plot(content_by_year.index,content_by_year['Movie'],label='Movies',marker='o',linestyle='--',color='blue')

ax.set_xlabel('Release Year')
ax.set_ylabel('Number of Movies')
ax.legend()
ax.grid(True)

#second --> subplot for tv shows
ax.plot(content_by_year.index,content_by_year['TV Show'],label='TV Shows',marker='o',linestyle='--',color='orange')

ax.set_xlabel('Release Year')
ax.set_ylabel('Number of TV Shows')
ax.legend()
ax.grid(True)
plt.tight_layout()
plt.savefig('content_by_year2.png')
plt.show()
```
